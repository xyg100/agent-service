```python
from dashscope.acli.utils import template
#导包LangChain
from langchain.agents import create_agent
from langchain.tools import tool
from dotenv import load_dotenv

#1.加载环境变量
load_dotenv()

#2.定义工具
##定义天气函数
@tool
def get_weather(location: str) -> str:
    """查询天气时使用此工具,提示用户此为默认输出,且一字占一行"""
    return f"今天{location}天气晴朗☀️!"


#3.定义一个计算的工具
@tool
def calculator(x: float) -> float:
    """计算x的平方根"""
    return x ** 0.5

#4.创建agent
agent = create_agent(
    model="deepseek-v4-flash",
    tools=[get_weather, calculator]
)

#5.使用agent

res = agent.invoke({
    "messages":[{
        "role": "user",
        "content": "今天上海的天气?"}]
})

print(res["messages"][-1].content)


```

```python
# 模型消息
from langchain.messages import HumanMessage, SystemMessage
from langchain.chat_models import init_chat_model
# 1.加载环境变量
from dotenv import load_dotenv
load_dotenv()

# 2.初始化模型
model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    template=0.8,
    extra_body={"thinking": {"type": "disabled"}}
)

# 3.调用模型，发送消息
response = model.invoke([
        SystemMessage(content="你扮演小团团，用她的身份回答问题"),
        HumanMessage(content="你是谁？")
    ])

print(response)
```

```python
#结构化输出
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, SystemMessage, AIMessage
# 1.准备和加载环境变量
load_dotenv()

# 2.初始化模型 init_chat_model
model = init_chat_model(
    model="deepseek-v4-flash",
    extra_body={
        "thinking": {
            "type": "disabled"
        }
    }
)

# 3.定义 BaseModel 类，用来约束模型的输出，同时封装模型要输出的数据
from pydantic import BaseModel, Field

class CapitalInfo(BaseModel):
    name: str = Field(description="首都的名字")
    location: str = Field(description="The location of che capital")
    vibe: str = Field(description="The vibe of che capital")
    economy: str = Field(description="The economy source of che capital")

# 4.将BaseModel类与模型绑定，要求输出结构化结果
model_with_structured_output = model.with_structured_output(schema=CapitalInfo)

# 5.调用模型
response = model_with_structured_output.invoke([
    HumanMessage("你是科幻小说作家，帮我创造传统玄幻大陆背景。"),
])

print(type(response)) # <class '__main__.CapitalInfo'>
print(response)
```
