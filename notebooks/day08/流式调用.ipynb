```python

# stream_events 调用Agent
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain.tools import tool


#1.加载配置
load_dotenv()
#2.初始化LLM客户端
model = init_chat_model(
    "deepseek-v4-flash",
    extra_body={"thinking": {"type": "disabled"}}
)

##定义天气函数
@tool
def get_weather(location: str) -> str:
    """Get the weather for a location.
       args:
           location (str): The location to check the weather for.
    """
    return f"It's sunny today in {location}!"


##定义一个计算的工具
@tool("calculator",
      description="A tool for calculating the square root of a number.")
def calculator(x: float) -> float:
    return x ** 0.5

#定义工具
tools = [get_weather, calculator]

#4.创建Agent
agent = create_agent(model = model,
                     tools = tools,
                     # system_prompt="请以海盗的口吻回答问题"
                     )

#5.流式调用
#5.1 调用api
stream = agent.stream_events({
        "messages": [HumanMessage(content="你好,介绍自己"),]
    },
    version="v3"
)
#5.2 解析结果并打印
for message in stream.messages:
    for content in message.text:
        print(content, end="|", flush=True)
```
