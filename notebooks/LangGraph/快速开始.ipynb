```python
from langgraph.constants import START, END
from langgraph.graph import StateGraph
from typing_extensions import TypedDict
#1. 定义 state：用 TypedDict 声明 dict 的键与类型；运行时字段只是类型标注，不是类属性
class MyState(TypedDict):
    input:str

#2. 定义 node：普通函数，入参是当前 state，返回"要更新的字段"（可为部分字段或 None）
#   START / END 是框架内置虚拟节点（'__start__' / '__end__'），不需要也不能注册
def node1(state:MyState) -> MyState:
    lower_input = state["input"].lower()
    return {"input":lower_input}

#3. 创建图构建器 StateGraph(state_schema=MyState)：此时还没有节点和边
graph_builder = StateGraph(state_schema=MyState)

#4.1 添加节点；
graph_builder.add_node("node1",node1)

# 4.2 添加边（START→node1→END）决定执行顺序
graph_builder.add_edge(START , "node1")
graph_builder.add_edge("node1" , END)

#5. 编译成可执行图，之后用 graph.invoke({...}) 传入初始 state 运行
graph = graph_builder.compile()

result = graph.invoke({"input":"ABC"})
print(result)


####################################构建图保存图片#########################################

# 1. 获取 PNG 二进制数据
png_data = graph.get_graph().draw_mermaid_png()

# 2. 以二进制写模式（"wb"）保存到本地文件
with open("graph_output.png", "wb") as f:
    f.write(png_data)

print("图表已成功保存为 graph_output.png")
```
